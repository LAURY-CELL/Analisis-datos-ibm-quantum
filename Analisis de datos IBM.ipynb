```python
import pandas as pd
import numpy as np
df = pd.read_csv('Datos IBM cusco.csv')
promedio_t1 = df['T1 (us)'].mean()
promedio_t2 = df['T2 (us)'].mean()
qubits_estables = df[(df['T1 (us)'] > promedio_t1) & (df['T2 (us)'] > promedio_t2)]
print(f"El promedio general de T1 es: {promedio_t1:.2f} us")
print(f"El promedio general de T2 es: {promedio_t2:.2f} us")
print(f"Número de qubits altamente estables encontrados: {len(qubits_estables)}")
print("\nLos 5 qubits con menor T1 (más ruidosos):")
print(df.nsmallest(5, 'T1 (us)')[['Qubit', 'T1 (us)']])
```
